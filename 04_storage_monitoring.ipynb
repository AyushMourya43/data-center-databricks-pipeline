```python

# DATA CENTER IoT - STORAGE MONITORING

base = "/Volumes/workspace/iot_db/data"

folders = [
    "dc_raw",
    "dc_bronze",
    "dc_silver",
    "dc_gold",
    "output",
    "logs"
]

def get_folder_size(path):
    total_size = 0

    try:
        files = dbutils.fs.ls(path)

        for file in files:
            if file.isDir():
                total_size += get_folder_size(file.path)
            else:
                total_size += file.size

    except Exception as e:
        print(f"Cannot access {path}: {e}")

    return total_size


grand_total = 0

print("DATA CENTER STORAGE REPORT")
print("-" * 45)

for folder in folders:

    path = f"{base}/{folder}"

    size_bytes = get_folder_size(path)
    size_mb = size_bytes / (1024 * 1024)
    size_gb = size_bytes / (1024 ** 3)

    grand_total += size_bytes

    print(f"{folder}: {size_mb:.2f} MB ({size_gb:.3f} GB)")


print("-" * 45)

print(
    f"TOTAL VOLUME STORAGE: "
    f"{grand_total / (1024 ** 3):.3f} GB"
)
```
